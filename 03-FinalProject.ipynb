## Abstract

College students commonly reduce sleep and maintain irregular bedtimes, but many prior studies rely on self-reports and do not jointly evaluate sleep duration and sleep regularity against academic outcomes. This project tests whether longer and more regular sleep is associated with higher academic performance, using objective sleep measurement as the primary evidence source.

Our primary dataset is the CMU Sleep Study (634 first-year college students from multiple universities), which combines Fitbit-derived sleep behavior with registrar-reported GPA. After quality filtering and missing-data handling, the analytic CMU sample contains 570 students. We use a supplementary Kaggle Sleep Health and Lifestyle dataset (synthetic; 155 retained rows after cleaning) only for contextual robustness checks on sleep-health relationships, not for direct GPA inference.

In the CMU data, sleep duration is positively associated with term GPA (r about 0.170), while bedtime irregularity is negatively associated with term GPA (r about -0.194). Group comparisons are directionally consistent with these associations: average GPA rises across sleep-duration groups from less than 6 hours to 8 or more hours. In the supplementary Kaggle data, longer sleep is strongly associated with better reported sleep quality (r about 0.917) and lower stress (r about -0.849), supporting broader sleep-health plausibility.

Overall, findings support the hypothesis directionally: longer and more regular sleep aligns with better outcomes in the primary college dataset. Effect sizes are modest for GPA, indicating sleep is one important factor among many. We therefore interpret results as meaningful but not deterministic, with explicit attention to confounding, generalizability limits, and ethics.

## Background and Prior Work

### Sleep and Cognitive Function

Sleep plays a fundamental role in maintaining cognitive health and academic performance. As sleep neuroscience research has established, sleep is essential for memory consolidation, attention restoration, and emotional regulation (Walker & Stickgold, 2006). The human body requires approximately 7-9 hours of sleep per night to function optimally, a window supported by expert consensus identifying recommended sleep durations as those "appropriate for health and well-being" (Hirshkowitz et al.). During sleep, the brain consolidates memories, processes emotional information, and restores metabolic resources necessary for wakeful cognitive performance (Goel, 2017).

Sleep deprivation has profound cognitive consequences. Research demonstrates that when individuals are sleep deprived, there are measurable declines in attention, working memory, and critical abilities for academic success. A meta analysis found that "sleep-deprived individuals showed consistent impairments across multiple cognitive domains, with deficits equivalent to or exceeding those observed with alcohol intoxication" (Lim & Dinges, 2010). For students specifically, this translates to reduced learning capacity, difficulty concentrating during lectures, and increased likelihood of academic errors on exams and assignments.

### Sleep Duration and Academic Performance

Research demonstrates a strong relationship between sleep duration and academic outcomes, particularly in college student populations. Multiple studies have found that college students who sleep fewer than 6 hours per night have significantly lower GPAs compared to their peers who maintain 7-9 hours of sleep nightly. In one study examining 148 college students, researchers found that "students sleeping less than 7 hours per night were significantly more likely to have below-average academic performance," with sleep duration accounting for meaningful variance in GPA (Giri et al., 2021). Moreover, research highlights that "one critical factor for academic success during the first year of college is students' sleep habits" (Creswell et al.).

The academic consequences show a linear decline. As sleep duration decreases, academic performance tends to decline correspondingly. This finding is particularly relevant to college populations, where sleep deprivation is often regarded as normal and even celebrated as a marker of dedication to studies. However, this cultural attitude toward sleep deprivation contrasts with the evidence that sleep is not a luxury but a biological necessity for optimal cognitive performance.

### Sleep Consistency and Regularity

Beyond total sleep duration, research increasingly emphasizes the importance of sleep consistency and regularity. As research has shown, "maintaining a regular sleep schedule is just as important for cognitive function as achieving adequate total sleep hours." This is because regular sleep patterns help synchronize circadian rhythms with behavioral patterns (Taub & Berger, 1973). Sleep irregularities such as varying bedtimes and wake times across days can disrupt these biological rhythms and compromise the restorative benefits of sleep, even when total duration is adequate.

Studies on circadian misalignment demonstrate that "irregular sleep schedules lead to desynchronization between the internal biological clock and external time cues, resulting in impaired cognitive performance despite nominally adequate sleep duration" (Monk & Folkard). This distinction between sleep quantity and sleep consistency is particularly important for studying college student populations. Students often face irregular schedules due to varying class times, midterms/finals periods, and social commitments. A student sleeping 9 hours but with highly variable sleep times may experience cognitive performance deficits similar to a student sleeping a consistent 7 hours nightly.

### Research Gap

While existing literature establishes the relationship between sleep patterns (both duration and consistency) and academic performance, most studies rely on self-reported data or limited sample sizes. Few studies simultaneously examine both sleep duration and sleep consistency as predictors of academic performance. This project aims to contribute to this research gap by analyzing datasets linking sleep patterns to academic records, while simultaneously examining how sleep consistency - often overlooked in favor of duration alone - predicts GPA. Understanding both dimensions of sleep behavior may provide more nuanced insights into how sleep affects academic success and could inform evidence-based recommendations for student health interventions.

### Dataset 1: CMU Sleep Study - Nightly Sleep Time and GPA in First-Year College Students

This is a multi-site empirical study examining the relationship between sleep patterns tracked via Fitbit wearables and academic performance in first-year college students. The data comes from five separate cohorts at three universities (CMU, University of Washington, and Notre Dame) studied across multiple years (2016-2019).

**Key Metrics:**
- **Total Sleep Time**: Average actual sleep time per night in minutes (continuous variable, typically 300-600 minutes / 5-10 hours). This is calculated from Fitbit data as the time spent in bed minus awake/restless periods. The recommended range is 420-540 minutes (7-9 hours) for college students.
- **Bedtime Mean Successive Squared Difference (MSSD)**: Measured in hours, this captures day-to-day variability in bedtime (when students go to sleep). Lower values indicate more regular sleep schedules; high variability disrupts circadian rhythms even with adequate total sleep.
- **Midpoint of Sleep**: Average midpoint between bedtime and wake time (in minutes after 11 pm). For example, 364 minutes means approximately 5:04 am. This helps characterize whether students are sleeping during typical night hours or shifted schedules.
- **Term GPA**: End-of-semester cumulative GPA on 0-4.0 scale. This is the primary outcome variable we are predicting.
- **Cumulative GPA (cum_gpa)**: GPA from previous semesters (typically just fall semester for first-years at 3 universities, fall+winter for UW quarters). Important control variable.
- **Term Units (Z-normalized)**: Number of course units/credits taken, normalized as z-scores within each university cohort since universities use different unit systems.
- **Demographics**: Gender (0=male, 1=female), Race (0=underrepresented minority, 1=non-underrepresented), First-gen status (0=non-first-gen, 1=first-gen). Allows testing for fairness and bias.

**Data Collection Methodology:**
- Students carried Fitbits for one month early in their spring academic term
- Sleep episodes identified using specific algorithm: episodes >=20 minutes separated by >=5 minutes
- "Main" sleep episode (longest between noon-to-noon) used as primary sleep measurement
- Grades obtained from university registrars
- 634 total participants across 5 study cohorts

**Data Strengths:**
- Objective, device-measured sleep data (Fitbit actigraphy), eliminating self-report bias
- Published peer-reviewed study with rigorous methodology (Creswell et al., 2023, PNAS)
- Multiple independent universities and time periods increase generalizability
- Large sample size providing statistical power
- Includes both sleep *duration* and sleep *variability* measures aligned with research questions
- Rich demographic data enables fairness and bias analysis

**Data Limitations/Concerns:**
- Limited to first-year college students only (may not generalize to upper-class students or non-traditional students)
- Temporal scope limited to one month early in spring semester (may not capture full semester patterns or seasonal effects)
- Some device non-compliance: users may not have worn Fitbit constantly (frac_nights_with_data variable available as quality indicator)
- Cumulative GPA from *previous* semesters, not concurrent with sleep tracking (introduces timing mismatch)
- May have survivor bias if students who dropped out are underrepresented
- Active tracking during study period may alter behavior (Hawthorne effect)

#### EDA Execution (Single Code Cell)

The following code cell performs the full EDA workflow, computes summary statistics and correlations, renders each figure directly in notebook output, and prints its corresponding discussion immediately after the figure.

```python
# Single-cell EDA pipeline: load data, compute summaries, and render all figures inline
from pathlib import Path
import warnings
import os

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

warnings.filterwarnings('ignore')
sns.set_theme(style='whitegrid', context='notebook')

cmu_path = Path('data/02-processed/cmu_sleep_processed.csv')
kaggle_path = Path('data/02-processed/kaggle_sleep_health_processed.csv')
output_dir = Path('results/eda')
output_dir.mkdir(parents=True, exist_ok=True)

if not cmu_path.exists():
    raise FileNotFoundError(f"Missing {cmu_path}. Generate processed data first.")

cmu = pd.read_csv(cmu_path)
required_cols = ['TotalSleepTime_hours', 'bedtime_mssd', 'term_gpa']
missing = [column for column in required_cols if column not in cmu.columns]
if missing:
    raise ValueError(f"CMU data missing required columns: {missing}")

# Create interpretable sleep groups for downstream analysis
cmu['sleep_duration_group'] = pd.cut(
    cmu['TotalSleepTime_hours'],
    bins=[0, 6, 7, 8, 24],
    labels=['<6h', '6-7h', '7-8h', '8+h'],
    include_lowest=True
)

def safe_cohens_d(group_a, group_b):
    a = np.asarray(group_a.dropna(), dtype=float)
    b = np.asarray(group_b.dropna(), dtype=float)
    if len(a) < 2 or len(b) < 2:
        return np.nan
    var_a = np.var(a, ddof=1)
    var_b = np.var(b, ddof=1)
    pooled_sd = np.sqrt(((len(a) - 1) * var_a + (len(b) - 1) * var_b) / (len(a) + len(b) - 2))
    if pooled_sd == 0:
        return np.nan
    return (np.mean(a) - np.mean(b)) / pooled_sd

print('CMU processed shape:', cmu.shape)
summary_cols = ['TotalSleepTime_hours', 'bedtime_mssd', 'term_gpa']
if 'cum_gpa' in cmu.columns:
    summary_cols.append('cum_gpa')
print('\nCMU summary statistics:')
print(cmu[summary_cols].describe().round(3))
print('\nCMU correlation matrix:')
print(cmu[['TotalSleepTime_hours', 'bedtime_mssd', 'term_gpa']].corr().round(3))

cmu_group_summary = (
    cmu.groupby('sleep_duration_group', observed=False)['term_gpa']
    .agg(['count', 'mean', 'median', 'std'])
    .reindex(['<6h', '6-7h', '7-8h', '8+h'])
    .round(3)
)
print('\nCMU GPA summary by sleep-duration group:')
print(cmu_group_summary)

low_sleep = cmu.loc[cmu['sleep_duration_group'] == '<6h', 'term_gpa']
high_sleep = cmu.loc[cmu['sleep_duration_group'] == '8+h', 'term_gpa']
mean_diff = high_sleep.mean() - low_sleep.mean()
d_value = safe_cohens_d(high_sleep, low_sleep)
print('\nCMU key comparison (<6h vs 8+h):')
print(f'Mean GPA difference (8+h minus <6h): {mean_diff:.3f}')
print(f'Cohen d style effect size (8+h vs <6h): {d_value:.3f}')

# Figure 1
print('\nFigure 1: CMU Sleep Duration Distribution')
fig, ax = plt.subplots(figsize=(8, 5))
sns.histplot(cmu['TotalSleepTime_hours'], bins=25, kde=True, color='#4C72B0', ax=ax)
ax.set_title('CMU: Distribution of Sleep Duration (hours)')
ax.set_xlabel('Sleep Duration (hours)')
ax.set_ylabel('Count')
fig.tight_layout()
fig.savefig(output_dir / 'figure_1_cmu_sleep_duration.png', dpi=200)
plt.show()
plt.close(fig)
print('Figure 1 Discussion: The distribution centers around roughly 6-7 hours, with comparatively fewer students in the 8+ hour range. This provides immediate context for why sleep deprivation may matter in this sample: many students are below commonly recommended sleep durations.')

# Figure 2
print('\nFigure 2: CMU Bedtime Irregularity (MSSD) Distribution')
fig, ax = plt.subplots(figsize=(8, 5))
sns.histplot(cmu['bedtime_mssd'], bins=25, kde=True, color='#55A868', ax=ax)
ax.set_title('CMU: Distribution of Bedtime Irregularity (MSSD)')
ax.set_xlabel('Bedtime MSSD (higher = less regular)')
ax.set_ylabel('Count')
fig.tight_layout()
fig.savefig(output_dir / 'figure_2_cmu_bedtime_mssd.png', dpi=200)
plt.show()
plt.close(fig)
print('Figure 2 Discussion: This plot shows the spread of bedtime variability. Higher MSSD values represent less regular sleep schedules; the right tail indicates a subset of students with substantially irregular bedtimes, which is consistent with the hypothesis that regularity can matter independently of duration.')

# Figure 3
print('\nFigure 3: CMU Sleep Duration vs Term GPA')
fig, ax = plt.subplots(figsize=(8, 5))
sns.regplot(
    data=cmu,
    x='TotalSleepTime_hours',
    y='term_gpa',
    scatter_kws={'alpha': 0.45, 's': 25},
    line_kws={'color': 'crimson', 'linewidth': 2},
    ax=ax
)
ax.set_title('CMU: Sleep Duration vs Term GPA')
ax.set_xlabel('Sleep Duration (hours)')
ax.set_ylabel('Term GPA')
fig.tight_layout()
fig.savefig(output_dir / 'figure_3_cmu_sleep_vs_gpa.png', dpi=200)
plt.show()
plt.close(fig)
print('Figure 3 Discussion: The fitted trend line is positive, indicating that longer sleep duration is associated with higher term GPA on average. While the spread shows sleep is not the only driver of GPA, the direction supports the project hypothesis.')

# Figure 4
print('\nFigure 4: CMU GPA by Sleep-Duration Group')
fig, ax = plt.subplots(figsize=(8, 5))
order = ['<6h', '6-7h', '7-8h', '8+h']
sns.boxplot(
    data=cmu,
    x='sleep_duration_group',
    y='term_gpa',
    order=order,
    palette='Set2',
    ax=ax
)
ax.set_title('CMU: Term GPA by Sleep Duration Group')
ax.set_xlabel('Sleep Duration Group')
ax.set_ylabel('Term GPA')
fig.tight_layout()
fig.savefig(output_dir / 'figure_4_cmu_group_gpa.png', dpi=200)
plt.show()
plt.close(fig)
print('Figure 4 Discussion: Group-based comparison makes the trend easier to interpret: mean and median GPA generally rise as sleep duration moves from <6h toward 8+h. This matches the directional expectation from the hypothesis and complements the regression-style trend in Figure 3.')

if kaggle_path.exists():
    kaggle = pd.read_csv(kaggle_path)
    print('\nKaggle processed shape:', kaggle.shape)
    kcols = [c for c in ['Sleep Duration', 'Quality of Sleep', 'Stress Level'] if c in kaggle.columns]
    if kcols:
        print('\nKaggle summary statistics:')
        print(kaggle[kcols].describe().round(3))
        print('\nKaggle correlation matrix:')
        print(kaggle[kcols].corr().round(3))

    if all(c in kaggle.columns for c in ['Sleep Duration', 'Quality of Sleep']):
        print('\nFigure 5: Kaggle Sleep Duration vs Sleep Quality (Supplementary)')
        fig, ax = plt.subplots(figsize=(8, 5))
        sns.regplot(
            data=kaggle,
            x='Sleep Duration',
            y='Quality of Sleep',
            scatter_kws={'alpha': 0.45, 's': 30},
            line_kws={'color': 'darkred', 'linewidth': 2},
            ax=ax
        )
        ax.set_title('Kaggle: Sleep Duration vs Sleep Quality')
        fig.tight_layout()
        fig.savefig(output_dir / 'figure_5_kaggle_sleep_vs_quality.png', dpi=200)
        plt.show()
        plt.close(fig)
        print('Figure 5 Discussion: In the supplementary dataset, longer sleep aligns with higher reported sleep quality. This is not direct evidence about GPA, but it is directionally consistent with the idea that sleep quantity relates to better outcomes.')

    if all(c in kaggle.columns for c in ['Sleep Duration', 'Stress Level']):
        print('\nFigure 6: Kaggle Sleep Duration vs Stress (Supplementary)')
        fig, ax = plt.subplots(figsize=(8, 5))
        sns.regplot(
            data=kaggle,
            x='Sleep Duration',
            y='Stress Level',
            scatter_kws={'alpha': 0.45, 's': 30},
            line_kws={'color': '#1f77b4', 'linewidth': 2},
            ax=ax
        )
        ax.set_title('Kaggle: Sleep Duration vs Stress')
        fig.tight_layout()
        fig.savefig(output_dir / 'figure_6_kaggle_sleep_vs_stress.png', dpi=200)
        plt.show()
        plt.close(fig)
        print('Figure 6 Discussion: This supplementary plot shows a negative trend between sleep duration and stress. It supports the broader sleep-health pattern and helps contextualize the CMU findings, while still not replacing CMU-based GPA conclusions.')

    if 'Sleep Duration' in kaggle.columns:
        print('\nFigure 7: Cross-Dataset Sleep Duration Comparison')
        fig, ax = plt.subplots(figsize=(8, 5))
        sns.kdeplot(cmu['TotalSleepTime_hours'], fill=True, label='CMU students', alpha=0.5, ax=ax)
        sns.kdeplot(kaggle['Sleep Duration'], fill=True, label='Kaggle population', alpha=0.4, ax=ax)
        ax.set_title('Sleep Duration Distribution: CMU vs Kaggle')
        ax.set_xlabel('Sleep Duration (hours)')
        ax.legend()
        fig.tight_layout()
        fig.savefig(output_dir / 'figure_7_cross_dataset_sleep_distribution.png', dpi=200)
        plt.show()
        plt.close(fig)
        print('Figure 7 Discussion: Both datasets are centered near 6-7 hours, but the shape differs across populations. This comparison is useful for robustness context and supports presenting CMU as the primary evidence source for academic outcomes.')
else:
    print('\nKaggle processed file not found; supplementary Kaggle figures skipped.')

print('\nEDA Summary:')
print('- CMU results show directionally consistent relationships: longer and more regular sleep aligns with higher GPA.')
print('- Group-level summaries and effect-size-style comparisons support the same directional conclusion.')
print('- Supplementary Kaggle plots reinforce broader sleep-health relationships.')
print('- Each figure now has its corresponding discussion printed immediately after generation.')

print(f"\nSaved archival copies of figures to: {output_dir.resolve()}")
print('Generated files:')
for fname in sorted(os.listdir(output_dir)):
    print('-', fname)
```

## Discussion and Conclusion

Across the primary CMU dataset, our evidence is directionally consistent: longer average sleep duration and lower bedtime irregularity are both associated with higher term GPA. This pattern appears in correlation estimates, regression-style trend visualizations, and group-level GPA comparisons, providing convergent support for the project hypothesis.

The magnitude of these relationships is meaningful but not deterministic. Academic performance is influenced by many factors, including prior preparation, course rigor, stress exposure, health conditions, and resource access. For that reason, sleep should be interpreted as an actionable contributor rather than a single-cause explanation of GPA variation.

Supplementary Kaggle analyses show strong associations between longer sleep, higher sleep quality, and lower stress. These findings are useful for contextual robustness, but because the dataset is synthetic and has no GPA variable, they are not used as direct evidence for academic-performance conclusions.

Limitations are important for accurate interpretation. The CMU sample is restricted to first-year students and captures sleep over a limited period, which constrains generalization across class years, institutions, and full academic cycles. As an observational analysis, the study cannot establish causality and may still contain unmeasured confounding.

Even with these constraints, the project contributes a clear, evidence-based conclusion: sleep duration and sleep regularity are practical, measurable correlates of academic outcomes in college students. A strong next step is a longitudinal or intervention-based design that tracks multiple semesters, incorporates richer confounder controls, and evaluates whether improving sleep regularity produces measurable GPA gains.
